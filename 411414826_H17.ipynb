```python
# 定義武器名稱與攻擊力資料
weapons = ["木棒", "手槍", "霰彈槍", "狙擊槍", "雷射劍"]
attacks = [10, 30, 50, 80, 100]

# 1. 列出所有武器的編號與名稱（從 1 開始編號）
for i in range(len(weapons)):
    print(f"{i + 1} {weapons[i]}")

# 2. 讓玩家輸入武器編號，查詢該武器的攻擊力
while True:
    choice = int(input("請輸入武器編號（0 結束）："))

    if choice == 0:
        break
    elif 1 <= choice <= len(weapons):
        # 索引值需要減 1
        name = weapons[choice - 1]
        atk = attacks[choice - 1]
        print(f"{name} 的攻擊力是 {atk}")
    else:
        print("找不到武器")

# 3. 顯示所有武器攻擊力的平均值
avg_attack = sum(attacks) / len(attacks)
print(f"平均攻擊力： {avg_attack}")
```
