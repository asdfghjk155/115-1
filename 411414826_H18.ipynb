```python
# 定義超市商品與價格
products = {"蘋果": 50, "香蕉": 30, "牛奶": 60, "麵包": 40}

# 1. 顯示超市商品選單
print("=== 超市商品 ===")
for name, price in products.items():
    print(f"{name}：{price}元")

# 建立一個字典用來記錄購買的商品與數量
cart = {}

# 2. 讓使用者輸入要購買的商品名稱，直到輸入 "/" 結束
while True:
    item = input("請輸入商品名稱（輸入 / 結束）：")

    if item == "/":
        break
    elif item in products:
        # 如果商品已在購物車中，數量加 1；否則初始化為 1
        cart[item] = cart.get(item, 0) + 1
    else:
        print("沒有這個商品，請重新輸入。")

# 3. 計算每個商品的購買數量、單項小計，並列出總金額
total_amount = 0
print("\n--- 購物明細 ---")
for name, count in cart.items():
    subtotal = products[name] * count
    total_amount += subtotal
    print(f"{name} x {count}，小計：{subtotal} 元")

print(f"總金額：{total_amount} 元")
```
