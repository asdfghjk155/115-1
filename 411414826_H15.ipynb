```python
# 建立一個函式來讀取輸入的清單
def get_warehouse_items(warehouse_name):
    items = []
    print(f"請輸入{warehouse_name}文具，輸入'/'結束：")
    while True:
        item = input().strip()
        if item == '/':
            break
        if item:  # 避免空白輸入
            items.append(item)
    return items

# 1. 輸入 A 倉與 B 倉的文具清單
warehouse_a = get_warehouse_items("A倉")
warehouse_b = get_warehouse_items("B倉")

# 2. 合併二倉清單並加上「膠水」
combined_list = warehouse_a + warehouse_b + ["膠水"]

# 3. 使用斜線 / 串接並輸出結果
result_str = "/".join(combined_list)
print("\n二倉合併清單：")
print(result_str)
```
