```python
# 建立一個空集合
backpack = set()
print("初始背包:", backpack)

# 1. 玩家撿到道具 "劍"、"盾"、"藥水"，用 add() 加入背包
backpack.add("劍")
backpack.add("盾")
backpack.add("藥水")
print("撿到道具後的背包:", backpack)

# 2. 玩家又撿到 "藥水"，再加入集合看看會發生什麼事？
backpack.add("藥水")
print("再次撿到藥水後的背包:", backpack)

# 3. 玩家使用 "盾"，從集合刪除
backpack.remove("盾")
print("使用盾後的背包:", backpack)

# 4. 玩家隨機掉落一個道具，用 pop() 顯示哪個道具被掉了
dropped_item = backpack.pop()
print("掉落的道具:", dropped_item)

# 5. 遊戲結束，顯示最後背包
print("遊戲結束，最後背包:", backpack)
```
