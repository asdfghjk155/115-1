```python
def get_input():
    """
    負責取得使用者輸入的單價 p 與數量 q
    """
    p = float(input("請輸入單價 p: "))
    q = float(input("請輸入數量 q: "))
    return p, q

def calculate_discount(original_amount):
    """
    負責計算原價打 9 折後的金額 (子函式)
    """
    return original_amount * 0.9

def process_order():
    """
    主處理函式：這裡展現了「副程式呼叫副程式」
    1. 呼叫 get_input() 取得 p 與 q
    2. 計算原價總金額
    3. 呼叫 calculate_discount() 計算 9 折金額
    """
    # 呼叫輸入函式
    p, q = get_input()

    # 計算原價總金額
    original_total = p * q

    # 呼叫打折計算函式（副程式呼叫副程式）
    discount_total = calculate_discount(original_total)

    # 輸出結果（使用 int() 讓金額呈現整數，符合範例格式）
    print(f"原價金額：{int(original_total)}元")
    print(f"9折金額：{int(discount_total)}元")

# --- 主程式 ---
if __name__ == "__main__":
    process_order()
```
