```python
try:
    username = input("請輸入帳號：")
    password = input("請輸入密碼：")

    # 檢查是否有輸入空值
    if not username.strip() or not password.strip():
        raise ValueError("帳號或密碼不可為空值")

except ValueError as e:
    # 捕捉空值或其他自訂錯誤
    print(e)
else:
    # 當沒有發生例外時執行（驗證帳號密碼是否正確）
    if username == "admin" and password == "1234":
        print("登入成功")
    else:
        print("登入失敗")
finally:
    # 不論成功或失敗，最後一定會執行的區塊
    print("系統結束")
```
