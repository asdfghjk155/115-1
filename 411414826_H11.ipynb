```python
import random

def attack():
    """
    隨機產生 10 到 30 點之間的攻擊傷害，並回傳該數值。
    """
    # random.randint(a, b) 可以產生包含 a 和 b 在內的隨機整數
    damage = random.randint(10, 30)
    return damage

# --- 主程式 ---
if __name__ == "__main__":
    # 呼叫 attack() 函式取得傷害值
    monster_damage = attack()

    # 顯示玩家對怪獸造成的傷害
    print(f"你攻擊了怪獸,造成{monster_damage}點傷害！")
```
