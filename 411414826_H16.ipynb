```python
# 1. 讀取使用者輸入的餐點
order = []
print("請輸入您要點的餐點，輸入'/'結束：")

while True:
    item = input().strip()
    if item == '/':
        break
    if item:  # 確保不加入空白輸入
        order.append(item)

# 2. 檢查是否有沙拉，若有則刪除並提示
if "沙拉" in order:
    order.remove("沙拉")
    print("今日無沙拉，已自動移除。")

# 3. 贈送冰淇淋
order.append("冰淇淋")

# 4. 列出完整訂單內容（用逗號與空格分隔）
print("完整訂單清單：")
print(", ".join(order))
```
